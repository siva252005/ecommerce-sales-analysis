```python
total_sales = df['Sales'].sum()
print(total_sales)
```

Insight:
- The total revenue generated is 2261536.7827
- This represents the overall business performance across all orders.


```python
region_sales = df.groupby('Region')['Sales'].sum()
print(region_sales)
```

Insight:
- West region has the highest sales.
- South region has the lowest sales.


```python
category_sales = df.groupby('Category')['Sales'].sum()
print(category_sales)
```

Insight:
- Technology category performs best.
- Other categories contribute moderately.

```python
monthly_sales = df.groupby('Month')['Sales'].sum()
monthly_sales
```

Insight:
- Highest sales in November, lowest in February.
- Sales increase towards the end of the year.

```python
top_products = df.groupby('Product Name')['Sales'].sum().sort_values(ascending=False).head(10)
top_products
```

Insight:
- Few top products generate major revenue.
- Office and tech products dominate sales.
